```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

print(load_dotenv())
```

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")

response = llm.invoke("How many moons does jupiter have?")

print(response.text)
```

```python
response = llm.invoke([
    ["system", "You are helpful assistant how gives answer in one short line"],
    ["human", "How many moons does jupiter have?"]
])

print(response.text)
```

```python
prompt = "Write a single tagline for a quick cofee shop. Only output the tagline, nothing else."

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", temperature=0)

response = llm.invoke(prompt)

print(response.text)
```

```python
prompt = "Write a single tagline for a quick cofee shop. Only output the tagline, nothing else."

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", temperature=1)

response = llm.invoke(prompt)

print(response.text)
```

#  Call LLM Using Groq Models

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b", reasoning_format="parsed")

messages = [
    ("system", "You are a helpful assistant that gives a short one line answer to a question"),
    ("human", "How many moons jupiter have?"),
]

ai_msg = llm.invoke(messages)
print(ai_msg.content)
```
