```python
import sys

from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
```

```python
with open ("blood_work.txt", "r") as blood:
    blood_report = blood.read()

print(blood_report[:200])
```

```python
extraction_prompt = f""""
You are a medical data extraction assistant.

From the blood report below, extract ALL test values and classify each one as HIGH, LOW, or NORMAL 
based on tghe reference renages provided in the report.

Format response as:
- Test Name: value | High/LOW/NORMAL | Reference: range

Blood Report:
{blood_report}
"""

extraction_response = llm.invoke(extraction_prompt)
exracted_values = extraction_response.text

print("=== STAGE 1: EXTRACTED VALUE ===")
print(exracted_values)
```
